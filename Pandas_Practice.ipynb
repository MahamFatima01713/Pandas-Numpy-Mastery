```python
import pandas as pd

# Load both datasets
matches = pd.read_csv('matches - matches.csv')
delivery = pd.read_csv('delivery.csv')

# View the first few rows of deliveries
delivery.head()
```

```python
matches.matches
```

```python
#Merge Function:
#Merge acts same as join funtion in SQL.It can merge two different dataset but their must be a link or u can say same thing
#so that they can link to each other
#First table name .merge
# 1. Load the CSV into a clean variable name 
match = pd.read_csv('matches - matches.csv')
delivery.merge(match, left_on='match_id', right_on='id')
```

```python
delivery.merge(match, left_on='match_id', right_on='id').shape
```

```python
print(delivery.shape)
print(match.shape)
```

```python
#Group by cant work on tuple.And by .shape this will become tuple
new=delivery.merge(match, left_on='match_id', right_on='id')# saving it in new then we will apply group by
new.groupby('batsman')['batsman_runs'].sum()
#('batsman')=>This acts as a string in this line.In parenthesis python will consider all the batsman coulmn as string
#['batsman_runs']=>By wrting this in [],python will consider it through all the datset and apply related function 
#on this coulmn of all rows
```

```python
#To pass two or more columns in gorupby u will create the list
#By this it will first create the group of season and batsman then will perform sum on the batsman_runs coulmn for all 
#rows.Will give season and batsman as index and sum of runs as value
new.groupby(['season','batsman'])['batsman_runs'].sum()
```

```python
#u want to find top batsman in every season
#First we made group of both batsman and season so that calculation can be perform on both
#then we do sum on batsman_runs and then sort runs in ascending order
new.groupby(['season','batsman'])['batsman_runs'].sum().sort_values(ascending=False)
```

```python
#Then we will apply reset_index (to convert series into dataframe).Through this it will now have index properly arrange
new.groupby(['season','batsman'])['batsman_runs'].sum().sort_values(ascending=False).reset_index()
```

```python
#Then we will drop all other values except first one ,because we arrange them in descending order and first one is
#highest that we actually want.We will keep season as subset cuz we want to drop duplicates based on season.
new.groupby(['season','batsman'])['batsman_runs'].sum().sort_values(ascending=False).reset_index().drop_duplicates(subset='season',keep='first')
#now we have batsman name with highest score from every season
```

```python
#now we want only batsman name and season so we will season and batsman name by[['season','batsman']]
new.groupby(['season','batsman'])['batsman_runs'].sum().sort_values(ascending=False).reset_index().drop_duplicates(subset='season',keep='first')[['season','batsman']]
#HEREEEEE WEEEEE AREEEEEEEE HEHEHEHE
```
